## Задача 1. Первая функция и `return`

*Новое: `def`, параметр, `return`, `None`*

Функцию создают при помощи ключевого слова `def`: имя, параметры в скобках, двоеточие, тело с отступом.
Функция, которая что-то **считает**, должна **возвращать** результат через `return`,
а не печатать его. Функция без `return` возвращает `None` - «ничего».

````python
def count_chars(text):
    return len(text)

def show_chars(text):
    print(len(text))

a = count_chars("кот спит")
b = show_chars("кот спит")      # напечатает 8, но вернёт None
print("a =", a, " b =", b)
print(count_chars("кот") + count_chars("пёс"))   # с результатом можно считать
````

## Задача 2. Функция, которая отвечает «да» или «нет»

*Новое: функция-предикат, ранний `return`*

Функцию, которая возвращает `True` или `False`, называют **предикатом**,
а имя дают как вопрос: `is_short`, `has_digit`.

`return` сразу завершает функцию. Этим удобно пользоваться в цикле:
нашли то, что искали, сразу выходим, дальше не смотрим.

````python
def is_short(text):
    return len(text.split()) < 4

def first_long_word(words, limit):
    for word in words:
        if len(word) > limit:
            return word      # нашли - сразу выходим
    return None              # дошли до конца - не нашли

print(is_short("кот спит"), is_short("кот спит на диване"))
print(first_long_word(["кот", "собака", "попугай"], 5))
print(first_long_word(["кот", "пёс"], 5))
````

**Ваша очередь:** напишите предикат `has_digit(text)`: идёт по символам и возвращает `True`
на первой же цифре (`ch.isdigit()`), а если цифр нет - `False`.
Проверьте на первом заголовке и на первом тексте, а потом выведите номера
всех текстов, в которых есть цифры.

**Должно получиться:**

```

тексты с цифрами: [0, 2, 3, 4, 7, 8]
```

````python
# Задача 2
def has_digit(text):
    for ch in text:
        if ch.isdigit():
            return True
    return False

print(has_digit(TITLES[0]), has_digit(TEXTS[0]))
indices = []
for i, text in enumerate(TEXTS):
    if has_digit(text):
        indices.append(i)
print('тексты с цифрами:', indices)
````

## Задача 3. Несколько значений и словарь

*Новое: `return a, b`, словарь как результат*

Функция может вернуть сразу два значения: `return a, b`. На самом деле это
кортеж, и его удобно сразу **распаковать**: `x, y = f()`.

Если значений много, лучше вернуть словарь: у каждого значения будет имя.

````python
def first_last(words):
    return words[0], words[-1]

first, last = first_last(["кот", "пёс", "ёж"])
print(first, last)

def word_card(word):
    return {"word": word, "length": len(word), "upper": word.upper()}

print(word_card("кот"))
````

  число слов среди текстов (используйте `word_count` из задачи 1);
- `news_card(i)` - возвращает словарь с ключами `"title"`, `"topic"`, `"words"`

## Задача 4. По умолчанию, по имени и с документацией

*Новое: значения по умолчанию, именованные аргументы, docstring*

У параметра может быть **значение по умолчанию**: `def f(x, width=20)`.
Тогда при вызове его можно не передавать. Аргумент можно передать **по имени**:
`f(text, width=40)` - так понятнее и порядок не важен.

Первая строка в тройных кавычках внутри функции - **документация**.
Её показывает `help(f)`.

````python
def repeat(word, times=2, sep=" "):
    """Повторяет слово times раз через разделитель sep."""
    return sep.join([word] * times)

print(repeat("ура"))
print(repeat("ура", 3))
print(repeat("ура", sep="-", times=4))
help(repeat)
````

**Ваша очередь:** напишите `shorten(text, width=30, tail="...")` с документацией: если текст

символов и `tail`. Вызовите четырьмя способами, как в образце и выведите `help`.

```
Help on function shorten:

shorten(text, width=30, tail='...')
    Обрезает текст до width символов и добавляет tail.
```

````python
# Задача 4
def shorten(text, width=30, tail="..."):
    """Обрезает текст до width символов и добавляет tail."""
    if len(text) <= width:
        return text
    return text[:width] + tail

print(shorten(TEXTS[0]))
print(shorten(TEXTS[0], 50))
print(shorten(TITLES[2]))
print(shorten(TEXTS[1], tail=' [далее]', width=20))
help(shorten)
````

## Задача 5. Функция вызывает функцию

*Новое: одна функция внутри другой*

Большую задачу делят на маленькие функции, и каждая делает одно действие.
`words` не чистит текст сама - она вызывает `clean`. Захотите обработать иначе -
поправите только `clean`.

````python
def clean(text):
    """Нижний регистр, знаки препинания заменены пробелами."""
    text = text.lower()
    for sign in ".,!?:;%":
        text = text.replace(sign, " ")
    return text

def words(text):
    """Список слов текста."""
    return clean(text).split()

print(words("Кот спит. Пёс - нет!"))
````

**Ваша очередь:** запустите разбор: `clean` и `words` понадобятся дальше. Напишите
`frequencies(items)` - частотный словарь `{элемент: сколько раз}` - и
`word_freq(text)`, которая вызывает `words` и `frequencies`. Посчитайте
частоты слов второго текста и напечатайте, сколько раз там встречаются «тренер»
и «клуб» и сколько в тексте разных слов. Потом примените `frequencies`
к рубрикам `TOPICS`: функции всё равно, что считать.

**Должно получиться:**

```
тренер: 2  клуб: 2  разных слов: 26
{'спорт': 2, 'экономика': 2, 'наука': 2, 'технологии': 2, 'Россия': 2}
```

````python
# Задача 5
def frequencies(items):
    freq = {}
    for item in items:
        freq[item] = freq.get(item, 0) + 1
    return freq

def word_freq(text):
    return frequencies(words(text))

freq = word_freq(TEXTS[1])
print('тренер:', freq.get('тренер', 0),
      ' клуб:', freq.get('клуб', 0), ' разных слов:', len(freq))
print(frequencies(TOPICS))
````

## Задача 6. Где живут переменные и чужие списки

*Новое: локальные переменные, `None` по умолчанию*

Переменная, созданная внутри функции, называется **локальной**: снаружи её нет.
Функция получает данные через параметры и возвращает через `return`, а глобальные
переменные не изменяет.

Список, переданный в функцию, не копируется: если функция выполнит `.remove()`,
изменится список того, кто её вызвал. Поэтому правило курса: **функция собирает
новый список и возвращает его**. По той же причине список по умолчанию пишут
как `None`, а не `[]`.

````python
def count_long(words, limit=5):
    count = 0                 # локальная: живёт только внутри функции
    for word in words:
        if len(word) > limit:
            count += 1
    return count

print(count_long(["кот", "собака", "попугай"]))
# print(count)                # NameError: снаружи count не существует

def add_word(word, bag=None):     # не bag=[]
    if bag is None:
        bag = []
    bag.append(word)
    return bag

print(add_word("кот"))
print(add_word("пёс"))            # новый список, а не ['кот', 'пёс']
````

## Задача 7. `*args` и `**kwargs`

`*args` собирает все позиционные аргументы в кортеж, `**kwargs` - все
именованные в словарь. Звёздочка при **вызове** работает наоборот:
`f(*список)` раскладывает список в отдельные аргументы.

````python
def longest(*words):
    best = ""
    for word in words:
        if len(word) > len(best):
            best = word
    return best

print(longest("кот", "собака", "ёж"))
print(longest(*["а", "бб", "в"]))

def show(**fields):
    for name, value in fields.items():
        print(f"  {name}: {value}")

show(рубрика="спорт", слов=34)
````

## Задача 8. Функция как значение: `key=` и `lambda`

*Новое: функция без скобок, `sorted(key=...)`, `lambda`*

Имя функции без скобок - это сама функция. Её можно передать в другую
функцию, например в `sorted(..., key=...)`: `sorted` вызовет её для каждого
элемента и отсортирует по результату.

Если функция нужна на одну строку, её пишут на месте: `lambda p: p[1]`.
Ключ из двух частей `(-p[1], p[0])` сортирует по числу вниз, а при равенстве -
по слову по алфавиту.

````python
def last_letter(word):
    return word[-1]

animals = ["кот", "ёж", "собака"]
print(sorted(animals, key=last_letter))
print(sorted(animals, key=len))

pairs = [("б", 2), ("а", 2), ("в", 5)]
print(sorted(pairs, key=lambda p: p[1], reverse=True))
print(sorted(pairs, key=lambda p: (-p[1], p[0])))
````

## Задача 9. Списковое включение, `map`, `filter`

Цикл, который собирает новый список, можно записать одной строкой:
`[f(x) for x in xs if условие]`. То же умеют встроенные `map` (применить
функцию к каждому) и `filter` (оставить те, для которых функция вернула `True`).
Им передают функцию - без скобок.

````python
animals = ["кот", "собака", "ёж", "попугай"]
print([len(w) for w in animals])
print([w for w in animals if len(w) > 3])
print(list(map(len, animals)))
print(list(filter(lambda w: len(w) > 3, animals)))
````

## Задача 10. Функция, которая возвращает функцию

*Новое: вложенная функция, замыкание*

Внутри функции можно определить другую функцию и вернуть её. Внутренняя
функция **запомнит** параметры внешней, даже когда та закончит свое выполнение.

````python
def multiplier(k):
    def multiply(x):
        return x * k
    return multiply

double = multiplier(2)
triple = multiplier(3)
print(double(5), triple(5))
````

## Шпаргалка занятия

| Что нужно | Как пишется |
| --- | --- |
| определить и вызвать | `def f(x): ...`, `f(5)` |
| вернуть результат | `return значение`; без `return` - `None` |
| ранний выход | `return` внутри цикла |
| несколько значений | `return a, b`, затем `x, y = f()` |
| словарь как результат | `return {"tokens": n, ...}` |
| по умолчанию / по имени | `def f(text, width=30)`, `f(text, width=50)` |
| документация | `"""Что делает."""` первой строкой, `help(f)` |
| не портить аргумент | собрать новый список и вернуть его |
| список по умолчанию | `bag=None`, внутри `if bag is None: bag = []` |
| сколько угодно аргументов | `*args` (кортеж), `**kwargs` (словарь), `f(*список)` |
| функция как значение | `sorted(xs, key=word_count)` |
| короткая функция | `lambda p: (-p[1], p[0])` |
| список одной строкой | `[f(x) for x in xs if cond]`, `map`, `filter` |
| функция возвращает функцию | `def outer(k): def inner(x): ...; return inner` |
| аннотации | `def f(text: str) -> int:` |
| остановить с ошибкой | `if not isinstance(x, str): raise TypeError("...")` |
| проверить функцию | `check(имя, получено, ожидалось)` |
